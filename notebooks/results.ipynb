## Audit the verifier: random answers marked correct / incorrect

```python
import json, random
run = "grpo_s1"
rows = [json.loads(l) for l in (root / run / "predictions.jsonl").read_text().splitlines()]
random.seed(0)
for row in random.sample(rows, 10):
    g = row["greedy"]
    print(row["unique_id"], "| gold:", row["gold"], "| pred:", g["predicted"], "| correct:", g["correct"], "|", g["finish"])
```
